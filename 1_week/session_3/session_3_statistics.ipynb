```python
def stratified_sampling(group_df):
  return group_df.sample(frac = 0.2)
```

## Distributions

```python
chances = ['H', 'T']
n_h = 0
n_t = 0
for i in range(1,100):
  result = np.random.choice(chances, 1)
  if(result == 'H'):
    n_h +=1
  else:
    n_t+=1
```
