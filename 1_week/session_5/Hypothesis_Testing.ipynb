```python
# A coffee shop claims that the average order value at their store is $15.
# Standard deviation: $3
# A market analyst wants to verify this claim. They randomly sampled 50 transactions and found:
# Mean order value: $16
```

```python
import numpy as np
u = 15
sigma = 3
x = 16
n = 50
# Imagine there are many such means
se = sigma/np.sqrt(n)

(x-u)/se
```

```python
# So coffee shop's claim is not correct
```

```python
# A public health department claims that 60% (0.60) of adults in a city get a flu shot every year.

# A local researcher suspects the actual coverage is lower.
# They take a random sample of 400 adults in the city and find that 220 of them received the flu shot.

# Test the researcher's claim at a 5% significance level (alpha = 0.05)
```

```python
p = 0.6

p_hat = 220/400

n = 400

se = np.sqrt(p*(1-p)/n)

z = (p_hat - p)/se

```

```python
z
```

```python
# An e-commerce company finds that historically, 10% (0.10) of website visitors add an item to their cart.

# The marketing team launches a new personalized recommendation widget on the homepage,
# believing it will increase the add-to-cart rate. They track a random sample of 900 visitors and find that 108
# of them add an item to their cart.

# Test the company's claim at a 8% significance level (alpha = 0.08).
```

```python
p = 0.1

p_hat = 108/900

n = 900

se = np.sqrt(p*(1-p)/n)

z = (p_hat - p)/se
1 - stats.norm.cdf(z)
```

```python
z
```

```python
# A digital marketing firm claims that its new optimization tool reduces the
# loading time of mobile landing pages to an average of 3.5 seconds.

# A website developer believes the actual average loading time is higher (slower) than what the firm claims.
# The developer randomly tests 16 landing pages and finds a sample mean loading time of 4.1 seconds
# with a population standard deviation of 1.2 seconds.

# Test the developer's claim at a 5% significance level (alpha = 0.05).  -- ONe tailed
```

```python
u = 3.5

sigma = 1.2


n = 16

x = 4.1

se = sigma/np.sqrt(n)

z_score = (x-u)/se
```

```python
z_score
```

```python
1- stats.norm.cdf(z_score)
```
