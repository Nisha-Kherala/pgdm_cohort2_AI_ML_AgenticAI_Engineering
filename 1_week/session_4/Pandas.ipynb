```python
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

# Generate sample data
data = {
    'ID': range(1, 21),  # 50 unique IDs
    'Name': ['Person_' + str(i) for i in range(1, 21)],  # Names from Person_1 to Person_50
    'Age': np.random.randint(18, 60, size=20),  # Random ages between 18 and 60
    'City': np.random.choice(['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'], size=20),
    'Score': np.random.uniform(0, 100, size=20)  # Random scores between 0 and 100
}

# Create the DataFrame
df = pd.DataFrame(data).head(20)
df
```
